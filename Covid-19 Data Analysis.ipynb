```python
covid_df.dropna(inplace = True)
```

```python
covid_df.drop(columns = ['Time','ConfirmedIndianNational','ConfirmedForeignNational'],inplace = True)
```

```python
covid_df.drop('Sno',inplace= True,axis = 1)
```

```python
covid_df.head()
```

```python
# Calculating the total Active Cases
covid_df['Active_Cases'] = covid_df['Confirmed']-(covid_df['Cured']+covid_df['Deaths'])
```

```python
statewise = pd.pivot_table(covid_df,values = ["Confirmed","Deaths","Cured"],index = "State/UnionTerritory",aggfunc = sum)
```

```python
statewise
```

```python
statewise['Recovery_Rate'] = (statewise['Cured']/statewise['Confirmed'])*100
```

```python
statewise['Mortality_Rate'] = (statewise['Deaths']/statewise['Confirmed'])*100
```

```python
statewise = statewise.sort_values(by='Confirmed',ascending = False)
```

```python
statewise
```

```python
statewise = statewise.drop('Unassigned')
```

```python
# top 10 active cases States
```

```python
top_10_active_cases = covid_df.groupby(by='State/UnionTerritory').max()[['Active_Cases','Date']].sort_values(by=['Active_Cases'],ascending = False).reset_index()
```

```python
top_10_active_cases
```

```python
fig = plt.figure(figsize = (16,9))
y = top_10_active_cases['Active_Cases'].iloc[:10]
x = top_10_active_cases['State/UnionTerritory'].iloc[:10]
plt.bar(x,y)
plt.xlabel('State')
plt.ylabel('Active Cases')
plt.title("Top 10 States according to Active State")
plt.show()
```

```python
# top 10 states with highest deaths
```

```python
top_10_death = covid_df.groupby(by = "State/UnionTerritory").max()[['Deaths','Date']].sort_values(by=['Deaths'],ascending = False).reset_index()
```

```python
fig = plt.figure(figsize = (16,9))
y = top_10_death['Deaths'].iloc[:10]
x = top_10_death['State/UnionTerritory'].iloc[:10]
plt.bar(x,y)
plt.xlabel('State')
plt.ylabel('Deaths')
plt.title("Top 10 States with Most Deaths")
plt.show()
```

```python
#Growth Trend
```

```python
fig = plt.figure(figsize=(12, 6))
states = ['Maharashtra', 'Karnataka', 'Kerala', 'Tamil Nadu', 'Uttar Pradesh']
filtered_df = covid_df[covid_df['State/UnionTerritory'].isin(states)]

# Plot the lineplot
ax = sns.lineplot(data=filtered_df, x='Date', y='Active_Cases', hue='State/UnionTerritory')
ax.set_title("Covid-19 Growth and Decline over the months")
plt.show()
```
